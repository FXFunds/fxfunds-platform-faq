# FXFunds Platform FAQ — Dataset

24 frequently asked questions about the FXFunds platform. Every answer is labelled with how it was verified, its source and the date it was last checked.

- Website: https://fxfunds.com/
- GitHub: https://github.com/FXFunds/fxfunds-platform-faq

*Informational only; not financial advice.*

```python
import glob
import pandas as pd

# Finds the CSV wherever Kaggle mounts the dataset
path = glob.glob('/kaggle/input/**/fxfunds_platform_faq.csv', recursive=True)[0]
df = pd.read_csv(path)
df[['id','category_name','question','status_label']].head(10)
```

## How answers are verified

```python
df['status_label'].value_counts()
```

## Questions per topic and status

```python
df.groupby(['category_name','status']).size().unstack(fill_value=0)
```

## What could not be located on official public pages

```python
df[df['status'] == 'not-located'][['id','question']]
```

## Simple keyword search

```python
def search(text):
    words = [w.lower() for w in text.split()]
    hits = df[df.apply(lambda r: all(w in (r['question'] + ' ' + r['answer']).lower() for w in words), axis=1)]
    return hits[['id','question','status_label']]

search('network fee')
```
